```python
import pandas as pd
import ast  # For safely evaluating string lists
import numpy as np

METHOD = ["reward_sim_eval_log_2000_syn.csv", "reward_sim_eval_log_2000_baseline.csv",
          "reward_sim_eval_log_4000_cluster_140_expQ2.csv", "reward_sim_eval_log_4000_cluster_140_none.csv",
          "reward_sim_eval_log_4000_cql_50.csv", "reward_sim_eval_log_rare_2000_cluster_120_expQ3.csv",
          "reward_sim_eval_log_rare_8000_cluster_100_expQ3.csv"]

for method in METHOD:
    file_path = f"reward_file_qualitative_analysis/{method}"

    FIRST_analysis = {'O_ACCEPTED': [], 'W_Assess_fraud': [], 'W_Fix_incomplete_submission': [], 'A_DECLINED': [],
                       'W_Call_after_offer': [], 
                       'W_Call_missing_information': []}

    df = pd.read_csv(file_path, sep=';')

    df['traces'] = df['traces'].apply(ast.literal_eval)

    # Iterate over the 'traces' column and print each list
    for trace in df['traces']:
        for act in FIRST_analysis:
            FIRST_analysis[act].append(trace.count(act))
    print("####################", method, len(FIRST_analysis['O_ACCEPTED']), "####################")
    for act in FIRST_analysis:
        print(act, np.mean(FIRST_analysis[act]))
```

```python
METHOD = ["reward_sim_eval_log_2000_syn.csv", "reward_sim_eval_log_2000_baseline.csv",
          "reward_sim_eval_log_4000_cluster_140_none.csv",
          "reward_sim_eval_log_4000_cluster_140_expQ2.csv", 
          "reward_sim_eval_log_4000_cql_50.csv", "reward_sim_eval_log_rare_2000_cluster_120_expQ3.csv",
          "reward_sim_eval_log_rare_8000_cluster_100_expQ3.csv"]

for method in METHOD:
    file_path = f"reward_file_qualitative_analysis/{method}"

    SECOND_analysis = {'Active_offer': [], 'W_Call_after_offer': [], 
                       'W_Call_missing_information': []}

    df = pd.read_csv(file_path, sep=';')

    df['traces'] = df['traces'].apply(ast.literal_eval)

    for trace in df['traces']:
        if 'O_ACCEPTED' in trace:
            SECOND_analysis['W_Call_missing_information'].append(trace.count('W_Call_missing_information'))
            SECOND_analysis['W_Call_after_offer'].append(trace.count('W_Call_after_offer'))
            SECOND_analysis['Active_offer'].append(trace.count('O_CREATED') - trace.count('O_CANCELLED'))
            
    #print(SECOND_analysis)
    print("####################", method, "####################")
    for act in SECOND_analysis:
        print(act, np.mean(SECOND_analysis[act]))
```
